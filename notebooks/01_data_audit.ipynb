```python
from pathlib import Path
import pandas as pd

# Supports running from either the project root or notebooks folder.
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

csv_path = (
    ROOT / "data" / "raw"
    / "四川省川剧院_时长及互动数据.csv"
)

df = pd.read_csv(csv_path, 
                 encoding="utf-8-sig",
                dtype={"作品ID":"string"}
                )

print("Rows and columns:", df.shape)
display(df.head())
display(df.isna().sum().to_frame("missing_count"))
```

```python
# Summarise the quality of each column
audit = pd.DataFrame({
    "data_type":df.dtypes.astype(str),
    "missing_count": df.isna().sum(),
    "missing_percent": df.isna().mean().mul(100).round(1),
    "unique_values": df.nunique()
})
display(audit)

# Check both full-row duplication and repeated post IDs
print("Exact duplicate rows:", df.duplicated().sum())
print("Repeated post IDs beyond the first:", df["作品ID"].duplicated().sum())

# Inspect content types and collection dates.
print("\nContent types:")
display(df["类型"].value_counts(dropna=False))

print("\nInteraction collection dates:")
display(df["互动采集日期"].value_counts(dropna=False))
```
